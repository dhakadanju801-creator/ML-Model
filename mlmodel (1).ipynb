```python
from ultralytics import YOLO
import cv2
import pyttsx3

# Initialize TTS
engine = pyttsx3.init()
engine.setProperty('rate', 150)  # speaking speed

# Load YOLO model
model = YOLO("runs/detect/stationery_yolov8/weights/best.pt")

# Open webcam
cap = cv2.VideoCapture(0)
spoken_objects = set()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Perform detection (no show=True)
    results = model(frame)

    # Draw and speak
    for r in results:
        boxes = r.boxes
        for box in boxes:
            cls = int(box.cls[0])
            conf = float(box.conf[0])
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            name = model.names[cls]

            # Draw rectangle and text
            cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(frame, f"{name} {conf:.2f}", (x1, y1 - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)

            # Voice output only once
            if name not in spoken_objects and conf > 0.5:
                spoken_objects.add(name)
                engine.say(f"{name} detected")
                engine.runAndWait()

    # Show fixed webcam window
    cv2.imshow("YOLO Detection with Voice", frame)

    # Quit if q pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
